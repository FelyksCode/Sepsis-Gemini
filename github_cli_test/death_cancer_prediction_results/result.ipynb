```python
import pandas as pd
import numpy as np
import os

# ====================================================
# Folder hasil iterasi
# ====================================================

base_path = ""

iterations = [
    "result_1",
    # "result_2",
    # "result_3"
]

models = [
    "medgemma",
    "mistral",
    "qwen3"
]

# ====================================================
# Simpan hasil gabungan
# ====================================================

all_results = []

for result_folder in iterations:

    for model in models:

        file_path = os.path.join(
            base_path,
            result_folder,
            f"llm_judge_results_{model}.csv"
        )

        df = pd.read_csv(file_path)

        df["iteration"] = result_folder
        df["model"] = model

        all_results.append(df)

combined_df = pd.concat(
    all_results,
    ignore_index=True
)

print("Total rows:")
print(combined_df.shape)

combined_df.head()
```

```python
combined_df.to_csv('combined_results.csv', index=False)
print("File saved as 'combined_results.csv'")
```

```python
# ====================================================
# Pilih kolom numerik
# ====================================================

metrics = [

    # SHAP
    "feature_accuracy",
    "direction_correctness",
    "shap_completeness",
    "shap_hallucination_count",
    "shap_final_score",

    # Clinical
    "shap_consistency",
    "clinical_plausibility",
    "clinical_relevance",
    "clinical_safety",
    "clinical_hallucination_count",
    "clinical_final_score"
]

# ====================================================
# Mean + Std per model
# ====================================================

summary = combined_df.groupby(
    "model"
)[metrics].agg(
    ["mean"]
)

summary.round(3)
```

```python
med = combined_df[combined_df["model"] == "medgemma"]

# counts
print("shap_hallucination_count == 1:", med["shap_hallucination_count"].eq(1).sum())
print("clinical_hallucination_count == 1:", med["clinical_hallucination_count"].eq(1).sum())

# show rows if any
print("\nRows with shap_hallucination_count == 1:")
print(med.loc[med["shap_hallucination_count"] == 1])

print("\nRows with clinical_hallucination_count == 1:")
print(med.loc[med["clinical_hallucination_count"] == 1])
```
